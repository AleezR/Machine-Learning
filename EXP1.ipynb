**EXPERIMENT 1 - IMPLEMENT LINEAR REGRESSION WITH ONE VARIABLE**
-Load dataset
-implement linier regression
-evaluate model performance

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#Load dataset
housing = fetch_california_housing()
```

```python
data = pd.DataFrame(housing.data,columns=housing.feature_names)
data["Price"]=housing.target
```

```python
#Select one feature
x = data[["AveRooms"]].values
y = data[["Price"]].values
```

```python
#Split data
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2, random_state=42 )
```

```python
#Feature Scaling
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
w = 0
b = 0
learning_rate = 0.1
epochs = 25
# Number of training samples
n = len(x_train_scaled)
for i in range(epochs):
    y_pred = w * x_train_scaled.flatten() + b
    dw = (1 / n) * np.sum((y_pred - y_train.flatten()) * x_train_scaled.flatten())
    db = (1 / n) * np.sum(y_pred - y_train.flatten())
    w = w - learning_rate * dw
    b = b - learning_rate * db
    if i % 5 == 0:
        cost = (1 / (2 * n)) * np.sum((y_pred - y_train.flatten()) ** 2)
        print(f"Epoch {i}, Cost = {cost:.4f}")
y_pred_gd = w * x_test_scaled.flatten() + b
```

```python
print("Gradient Descent")
print("------------------")
print("Weight:",w)
print("Bias:",b)
print("MSE:",mean_squared_error(y_test,y_pred_gd))
print("R2 Score:",r2_score(y_test,y_pred_gd))
```
